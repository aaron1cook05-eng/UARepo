# Challenge 4:
Write the python code to find the factorial of a number using a loop. Thefactorial of a number (integer positive) is the sum of multiplication of all the integers smaller than that positive integer. For example, factorial of 5 is:

5 * 4 * 3 * 2 * 1 which equals to 120.

How you can do that in a python script?, please try your best without using google.

```python
factorial = 5
factorial_result = 1
 
#designated the factorial number desired (5), this will influence the amount of integers the loop processes,
#"factorial_result" #1 stores the previous result so it can be multiplied by the next.

for j in range(1, factorial + 1):
    factorial_result*= j
    print(f"Factorial result: {factorial_result}")
    
#Code used to run the loop (for j in range) includes the factorial (5) and the result (1), 
#I used +1 after (5) as to ensure pythons range includes the fifth as without that the range would drop to 1-4. 

#'*' multilpies the integers together to calculate the factorials of (5). 

#Finishing with printing the intermediate and final factorial values.
```

```python
sensor_zones.groupby(
    by="WZCD",
    as_index=True
)

#Group by working zones as desired, keeping WZCD as a column.
```

```python
sensor_counts = sensor_zones.groupby("WZCD").size().reset_index(name="sensor_count")

#Group sensors by working zone (WZCD) and count the sensors in each zone.

#Reset the index and name the resulting count column 'sensor_count'.

sensor_counts.head()
```

```python
working_zones = working_zones.merge(
    sensor_counts,
    on="WZCD",
    how="left"
)

# Join the sensor counts to the working zones using WZCD as the common field.

# Keep all working zones, including those without a matching sensor count.
```
